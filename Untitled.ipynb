```python
def best_first_search(graph, start, goal, heuristic):
    open_list = [(start, [start])]   # (node, path)
    visited = []
    nodes_explored = 0

    start_time = 0  #
    while open_list:
        min_index = 0
        for i in range(len(open_list)):
            if heuristic[open_list[i][0]] < heuristic[open_list[min_index][0]]:
                min_index = i

        current, path = open_list.pop(min_index)
        nodes_explored += 1

        if current == goal:
            return path, start_time, nodes_explored
        if current not in visited:
            visited.append(current)
            for neighbor in graph[current]:
                if neighbor not in visited:
                    open_list.append((neighbor, path + [neighbor]))

    return None, start_time, nodes_explored
graph = {'A': ['B', 'C'], 'B': ['D', 'E'],'C': ['F'],'D': [], 'E': ['F'], 'F': []}

heuristic = {'A': 5, 'B': 4,'C': 2, 'D': 7,'E': 1,'F': 0}

path, time_taken, nodes = best_first_search(graph, 'A', 'F', heuristic)

print("Best-First Search:")
print("Path:", path)
print("Time taken (approx):", time_taken)
print("Nodes explored:", nodes)
```

```python
def a_star_search(graph, start, goal, heuristic):
    open_list = [(start, [start], 0)]  # (node, path, g_cost)
    visited = {}
    nodes_explored = 0
    start_time = 0  
    while open_list:
        min_index = 0
        for i in range(len(open_list)):
            node, path, g = open_list[i]
            f = g + heuristic[node]
            node_min, path_min, g_min = open_list[min_index]
            f_min = g_min + heuristic[node_min]
            if f < f_min:
                min_index = i
        current, path, g = open_list.pop(min_index)
        nodes_explored += 1

        if current == goal:
            return path, g, start_time, nodes_explored
        if current not in visited or visited[current] > g:
            visited[current] = g

            for neighbor, cost in graph[current]:
                new_g = g + cost
                open_list.append((neighbor, path + [neighbor], new_g))

    return None, None, start_time, nodes_explored

graph = {'A': [('B', 1), ('C', 4)],'B': [('D', 2), ('E', 5)],'C': [('F', 1)],'D': [],'E': [('F', 1)],'F': []}

heuristic = {'A': 7, 'B': 6,'C': 2,'D': 1,'E': 1, 'F': 0}

path, cost, time_taken, nodes = a_star_search(graph, 'A', 'F', heuristic)

print("\nA* Search:")
print("Path:", path)
print("Path Cost:", cost)
print("Time taken (approx):", time_taken)
print("Nodes explored:", nodes)
```
